## 6. Missing Value Analysis

### Missing Data Handling

The original Student Performance dataset contained no naturally occurring missing values. To demonstrate the required statistical imputation technique, 20 missing values were intentionally simulated in the `absences` variable using a fixed random seed.

Median imputation was then applied to replace the simulated missing values. After imputation, the number of missing values in `absences` was reduced to zero.

```python
missing = df.isnull().sum()

print("Total missing values:", missing.sum())

print("\nColumns with missing values:")
print(missing[missing > 0])
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage[missing_percentage > 0])
```

```python
df_missing = df.copy()
```

```python
np.random.seed(42)

missing_indices = np.random.choice(
    df_missing.index,
    size=20,
    replace=False
)

df_missing.loc[missing_indices, "absences"] = np.nan
```

```python
df_missing["absences"].isnull().sum()
```

```python
df_missing.isnull().sum()
```

## 7. Statistical Imputation

```python
median_absences = df_missing["absences"].median()

print("Median absences:", median_absences)
```

```python
df_missing["absences"] = df_missing["absences"].fillna(
    median_absences
)
```

```python
print(
    "Missing absences after imputation:",
    df_missing["absences"].isnull().sum()
)
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["absences"])

plt.title("Boxplot of Student Absences Before Outlier Treatment")
plt.xlabel("Absences")
plt.show()
```

## 8. Outlier Detection Using IQR

```python
Q1 = df["absences"].quantile(0.25)
Q3 = df["absences"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

## 9. Outlier Treatment

### Outlier Detection Results

The Interquartile Range (IQR) method was applied to the `absences` variable.

- Q1 = 0
- Q3 = 8
- IQR = 8
- Lower Bound = -12
- Upper Bound = 20
- Number of detected outliers = 15

The extreme absence values were capped at the upper IQR boundary of 20.

```python
outliers = df[
    (df["absences"] < lower_bound) |
    (df["absences"] > upper_bound)
]

print("Number of outliers:", len(outliers))
```

```python
outliers["absences"].sort_values()
```

```python
df_clean = df.copy()
```

```python
df_clean["absences"] = np.clip(
    df_clean["absences"],
    lower_bound,
    upper_bound
)
```

```python
print("Maximum absences before treatment:", df["absences"].max())
print("Maximum absences after treatment:", df_clean["absences"].max())
```

```python
outliers_after = df_clean[
    (df_clean["absences"] < lower_bound) |
    (df_clean["absences"] > upper_bound)
]

print("Outliers after treatment:", len(outliers_after))
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df_clean["absences"])

plt.title("Boxplot of Student Absences After IQR Outlier Treatment")
plt.xlabel("Absences")
plt.show()
```

## 10. Feature Engineering

### Engineered Features

Three new predictive features were created:

1. **Average_Grade** — the average of G1, G2, and G3.
2. **Grade_Improvement** — the difference between the final grade (G3) and the first-period grade (G1).
3. **Study_Absence_Ratio** — study time divided by absences plus one, preventing division by zero.

```python
df_clean["Average_Grade"] = (
    df_clean["G1"] + df_clean["G2"] + df_clean["G3"]
) / 3
```

```python
df_clean[["G1", "G2", "G3", "Average_Grade"]].head()
```

```python
df_clean["Grade_Improvement"] = (
    df_clean["G3"] - df_clean["G1"]
)
```

```python
df_clean[["G1", "G3", "Grade_Improvement"]].head()
```

```python
df_clean["Study_Absence_Ratio"] = (
    df_clean["studytime"] /
    (df_clean["absences"] + 1)
)
```

```python
df_clean[
    ["studytime", "absences", "Study_Absence_Ratio"]
].head()
```

```python
new_features = [
    "Average_Grade",
    "Grade_Improvement",
    "Study_Absence_Ratio"
]

df_clean[new_features].describe()
```

## Project Summary

The Student Performance dataset was analyzed and transformed through a structured data-cleaning and feature-engineering process.

The original dataset contained 395 records and 33 columns with no naturally occurring missing values. A controlled missing-value experiment was performed on the `absences` variable, and the simulated missing values were successfully handled using median imputation.

Outliers in the `absences` variable were identified using the Interquartile Range (IQR) method. Fifteen outliers were detected, with an upper boundary of 20. IQR capping was then applied to neutralize extreme values.

Three new predictive features were engineered:
1. Average_Grade
2. Grade_Improvement
3. Study_Absence_Ratio

The final dataset contains 395 records and 36 columns and is prepared for further machine learning analysis.
